# Visualizing the electric field of a point charge using Python and matplotlib
We wish to visualize the field of a 1C point charge located at the origin (0,0) m. 
$$ \vec{E} = \frac{k q}{r^2} \hat{r}, $$
where $\vec{E}$ is the electric field vector, $k=\qty{8.99e9}{\newton\meter\squared\per\coulomb\squared}$ is Coulomb's constant, and $q=\qty{1}{\coulomb}$ is the charge. 

```python
import numpy as np
import matplotlib.pyplot as plt

k = 8.99e9      
q = 1e-6        
N = 25        

x = np.linspace(0.0, 12.0, N) + 1e-6
y = np.linspace(0.0, 14.0, N) + 1e-6
X, Y = np.meshgrid(x, y)

xq, yq = 6.0, 7.0

dx = X - xq
dy = Y - yq

r2 = dx**2 + dy**2
thE = np.arctan2(dy, dx)

magE = k * q / r2

mask = magE < 1e6 

Ex = magE * np.cos(thE)
Ey = magE * np.sin(thE)

fig, ax = plt.subplots(figsize=(8, 8))
ax.quiver(X[mask], Y[mask], Ex[mask], Ey[mask], color='navy')

ax.plot(xq, yq, 'ro', markersize=10, label='wow')

ax.set_aspect('equal')
ax.set_xlabel('x (cm)')
ax.set_ylabel('y (cm)')
ax.set_title('67singlecharge')
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(loc='upper left')


plt.savefig("singlecharge", dpi=300, bbox_inches='tight')


plt.show()
```
